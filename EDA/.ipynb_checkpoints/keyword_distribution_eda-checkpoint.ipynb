# Google Trends Keywords Distribution EDA

This notebook explores the distributional properties of all keyword search interest data across categories and time. Run `stitch_keywords.py` first to generate `all_keywords_long.csv`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid', font_scale=1.1)
plt.rcParams['figure.dpi'] = 120
plt.rcParams['figure.facecolor'] = 'white'
```

```python
# Load stitched data
df = pd.read_csv('../Data/Intermediate/all_keywords_long.csv', parse_dates=['date'])
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month

print(f"Shape: {df.shape}")
print(f"Date range: {df['date'].min().date()} → {df['date'].max().date()}")
print(f"Categories: {df['category'].nunique()}")
print(f"Keywords: {df['keyword'].nunique()}")
print(f"\nNulls in search_interest: {df['search_interest'].isna().sum()} ({df['search_interest'].isna().mean():.2%})")
df.head()
```

## 1. Overall Distribution of Search Interest

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Histogram
axes[0].hist(df['search_interest'].dropna(), bins=50, color='steelblue', edgecolor='white', alpha=0.8)
axes[0].set_title('Histogram of All Search Interest Values')
axes[0].set_xlabel('Search Interest (0–100)')
axes[0].set_ylabel('Frequency')

# KDE
sns.kdeplot(df['search_interest'].dropna(), ax=axes[1], fill=True, color='steelblue')
axes[1].set_title('KDE of All Search Interest Values')
axes[1].set_xlabel('Search Interest (0–100)')

# Log-transformed histogram (for heavy-tailed view)
log_vals = np.log1p(df['search_interest'].dropna())
axes[2].hist(log_vals, bins=50, color='darkorange', edgecolor='white', alpha=0.8)
axes[2].set_title('Histogram of log(1 + Search Interest)')
axes[2].set_xlabel('log(1 + Search Interest)')
axes[2].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

Useful for identifying high-variance vs stable keywords.

```python
summary = df.groupby(['category', 'keyword'])['search_interest'].agg(
    mean='mean', median='median', std='std',
    skew='skew', min='min', max='max',
    pct_zero=lambda x: (x == 0).mean()
).round(2).sort_values(['category', 'keyword']).reset_index()

# Coefficient of variation
summary['cv'] = (summary['std'] / summary['mean']).round(3)

print(f"Keywords with highest volatility (CV):")
display(summary.nlargest(15, 'cv')[['category', 'keyword', 'mean', 'std', 'cv', 'skew']])

print(f"\nKeywords with highest skew:")
display(summary.nlargest(15, 'skew')[['category', 'keyword', 'mean', 'skew', 'pct_zero']])
```

## 9. Distribution Shape Classification
Classifies each keyword's distribution type (normal-ish, right-skewed, left-skewed, bimodal, zero-heavy) to guide transformations.

```python
def classify_distribution(series):
    """Simple heuristic classification of a keyword's distribution shape."""
    s = series.dropna()
    if len(s) < 10:
        return 'insufficient_data'
    
    pct_zero = (s == 0).mean()
    skewness = s.skew()
    kurt = s.kurtosis()
    
    if pct_zero > 0.3:
        return 'zero-heavy'
    elif abs(skewness) < 0.5:
        return 'roughly symmetric'
    elif skewness > 1.0:
        return 'strongly right-skewed'
    elif skewness > 0.5:
        return 'moderately right-skewed'
    elif skewness < -1.0:
        return 'strongly left-skewed'
    elif skewness < -0.5:
        return 'moderately left-skewed'
    else:
        return 'other'

classifications = df.groupby(['category', 'keyword'])['search_interest'].apply(
    classify_distribution
).reset_index(name='dist_type')

# Summary
print("Distribution shape counts across all keywords:")
print(classifications['dist_type'].value_counts())

fig, ax = plt.subplots(figsize=(14, 6))
ct = pd.crosstab(classifications['category'], classifications['dist_type'])
ct.index = [c.replace('_', ' ') for c in ct.index]
ct.plot.bar(stacked=True, colormap='Set2', ax=ax, edgecolor='white')
ax.set_title('Distribution Shape Classification by Category')
ax.set_ylabel('Number of Keywords')
ax.set_xlabel('')
ax.legend(title='Shape', bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=9)
ax.tick_params(axis='x', rotation=35)
plt.tight_layout()
plt.show()
```

# Comparing Keyword Search Behaviour During and Outside of Recession months

```python
# Load NBER recession indicator
rec = pd.read_csv('../Data/Intermediate/USREC_cleaned.csv')
rec.columns = ['date', 'recession']
rec['date'] = pd.to_datetime(rec['date'])
rec['recession'] = rec['recession'].astype(int)

# Merge with keyword data
df_rec = df.merge(rec, on='date', how='inner')
df_rec['recession_label'] = df_rec['recession'].map({0: 'Expansion', 1: 'Recession'})

print(f"Months in recession: {rec['recession'].sum()} / {len(rec)} ({rec['recession'].mean():.1%})")
print(f"Keyword-months in recession: {df_rec['recession'].sum():,} / {len(df_rec):,}")
```

```python
cat_rec_means = df_rec.groupby(['category', 'recession_label'])['search_interest'].mean().unstack()
cat_rec_means['pct_change'] = ((cat_rec_means['Recession'] - cat_rec_means['Expansion']) 
                                / cat_rec_means['Expansion'] * 100).round(1)
cat_rec_means = cat_rec_means.sort_values('pct_change', ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(18, 6))

# Grouped bar chart
cat_rec_means[['Expansion', 'Recession']].plot.barh(ax=axes[0], color=['steelblue', 'firebrick'])
axes[0].set_title('Mean Search Interest: Recession vs Expansion')
axes[0].set_xlabel('Mean Search Interest')
axes[0].set_ylabel('')
axes[0].legend(title='Period')

# Pct change
colors = ['firebrick' if x > 0 else 'steelblue' for x in cat_rec_means['pct_change']]
axes[1].barh(cat_rec_means.index, cat_rec_means['pct_change'], color=colors)
axes[1].axvline(0, color='black', linewidth=0.8)
axes[1].set_title('% Change in Mean Search Interest During Recessions')
axes[1].set_xlabel('% Change')
for i, v in enumerate(cat_rec_means['pct_change']):
    axes[1].text(v + (1 if v > 0 else -1), i, f'{v:+.1f}%', va='center', fontsize=9)

plt.tight_layout()
plt.show()

cat_rec_means
```

However, this can be flawed since the average number of searches become higher over time, it can't represent all of expansion. I could potentially look specifically between a time window. 

```python
kw_rec = df_rec.groupby(['category', 'keyword', 'recession_label'])['search_interest'].agg(
    ['mean', 'median', 'std']
).unstack('recession_label')

# Flatten multi-index columns
kw_rec.columns = [f'{stat}_{period}' for stat, period in kw_rec.columns]
kw_rec = kw_rec.reset_index()

# Recession lift (% increase in mean during recession)
kw_rec['recession_lift_pct'] = (
    (kw_rec['mean_Recession'] - kw_rec['mean_Expansion']) / kw_rec['mean_Expansion'].clip(lower=0.1) * 100
).round(1)

# Effect size (Cohen's d approximation)
kw_rec['cohens_d'] = (
    (kw_rec['mean_Recession'] - kw_rec['mean_Expansion']) / 
    ((kw_rec['std_Expansion'].fillna(1) + kw_rec['std_Recession'].fillna(1)) / 2)
).round(3)

print("=== TOP 25 Keywords with HIGHEST recession lift ===")
top_up = kw_rec.nlargest(25, 'recession_lift_pct')[
    ['category', 'keyword', 'mean_Expansion', 'mean_Recession', 'recession_lift_pct', 'cohens_d']
]
display(top_up)

print("\n=== TOP 15 Keywords that DECREASE during recessions ===")
top_down = kw_rec.nsmallest(15, 'recession_lift_pct')[
    ['category', 'keyword', 'mean_Expansion', 'mean_Recession', 'recession_lift_pct', 'cohens_d']
]
display(top_down)
```

```python
top_kws = kw_rec.nlargest(12, 'cohens_d')[['category', 'keyword']]
bottom_kws = kw_rec.nsmallest(6, 'cohens_d')[['category', 'keyword']]
selected = pd.concat([top_kws, bottom_kws])

fig, axes = plt.subplots(3, 6, figsize=(24, 12))
axes = axes.flatten()

for i, (_, row) in enumerate(selected.iterrows()):
    subset = df_rec[(df_rec['category'] == row['category']) & (df_rec['keyword'] == row['keyword'])]
    sns.violinplot(data=subset, x='recession_label', y='search_interest',
                   palette={'Expansion': 'steelblue', 'Recession': 'firebrick'},
                   ax=axes[i], inner='quartile', cut=0)
    axes[i].set_title(f"{row['keyword'].replace('_', ' ')}\n({row['category'].replace('_', ' ')})", fontsize=8)
    axes[i].set_xlabel('')
    axes[i].set_ylabel('')

for j in range(len(selected), len(axes)):
    axes[j].set_visible(False)

fig.suptitle('Recession vs Expansion: Top Sensitive Keywords (top and bottom row)', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Pick top 3 recession-rising and top 3 recession-falling by effect size
top3_up = kw_rec.nlargest(3, 'cohens_d')
top3_down = kw_rec.nsmallest(3, 'cohens_d')
showcase = pd.concat([top3_up, top3_down])

fig, axes = plt.subplots(2, 3, figsize=(20, 8))
axes = axes.flatten()

for i, (_, row) in enumerate(showcase.iterrows()):
    kw_ts = df_rec[(df_rec['category'] == row['category']) & (df_rec['keyword'] == row['keyword'])]
    ax = axes[i]
    ax.plot(kw_ts['date'], kw_ts['search_interest'], color='steelblue', linewidth=1)
    ax.set_title(f"{row['keyword'].replace('_', ' ')}\n(lift: {row['recession_lift_pct']:+.0f}%)", fontsize=9)
    ax.set_ylabel('Search Interest')
    
    # Shade recession periods
    rec_months = rec[rec['recession'] == 1]['date']
    for d in rec_months:
        ax.axvspan(d, d + pd.DateOffset(months=1), alpha=0.15, color='red')

fig.suptitle('Time Series with Recession Shading (red) — Top Rising & Falling Keywords', fontsize=13)
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import ttest_ind, mannwhitneyu

test_results = []
for (cat, kw), grp in df_rec.groupby(['category', 'keyword']):
    exp = grp.loc[grp['recession'] == 0, 'search_interest'].dropna()
    rec_vals = grp.loc[grp['recession'] == 1, 'search_interest'].dropna()
    
    if len(rec_vals) < 5:
        continue
    
    t_stat, t_pval = ttest_ind(rec_vals, exp, equal_var=False)
    u_stat, u_pval = mannwhitneyu(rec_vals, exp, alternative='two-sided')
    
    test_results.append({
        'category': cat, 'keyword': kw,
        'mean_expansion': exp.mean(), 'mean_recession': rec_vals.mean(),
        't_stat': round(t_stat, 3), 't_pval': t_pval,
        'u_stat': u_stat, 'u_pval': u_pval
    })

tests = pd.DataFrame(test_results)
tests['t_pval_adj'] = np.minimum(tests['t_pval'] * len(tests), 1.0)  # Bonferroni
tests['significant_bonf'] = tests['t_pval_adj'] < 0.05

print(f"Keywords significantly different during recessions (Bonferroni α=0.05): "
      f"{tests['significant_bonf'].sum()} / {len(tests)}")

print("\n=== Most significant recession-sensitive keywords ===")
display(tests.nsmallest(20, 't_pval')[
    ['category', 'keyword', 'mean_expansion', 'mean_recession', 't_stat', 't_pval', 't_pval_adj']
].round(6))
```

```python
kw_rec_class = kw_rec.merge(classifications, on=['category', 'keyword'])

pivot_lift = kw_rec_class.groupby(['category', 'dist_type'])['recession_lift_pct'].median().unstack()
pivot_lift.index = [c.replace('_', ' ') for c in pivot_lift.index]

fig, ax = plt.subplots(figsize=(14, 7))
sns.heatmap(pivot_lift, cmap='RdBu_r', center=0, annot=True, fmt='.0f',
            linewidths=0.5, cbar_kws={'label': 'Median Recession Lift (%)'}, ax=ax)
ax.set_title('Median Recession Lift (%) by Category × Distribution Type')
ax.set_ylabel('')
plt.tight_layout()
plt.show()
```

# SECTION 13: CORRECTED ANALYSIS — DETRENDED KEYWORDS vs ECONOMIC INDICATORS

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# ── Load all economic data with consistent date parsing ─────────────────────

def load_econ(path, date_col, value_cols=None):
    d = pd.read_csv(path)
    d[date_col] = pd.to_datetime(d[date_col])
    d = d.rename(columns={date_col: 'date'})
    if value_cols:
        d = d[['date'] + value_cols]
    return d

rec   = load_econ('../Data/Intermediate/USREC_cleaned.csv', 'observation_date', ['USREC'])
gdp   = load_econ('../Data/Intermediate/GDP_YoY_pct_cleaned.csv', 'date', ['gdp_yoy_pct_change'])
urate = load_econ('../Data/Intermediate/unemp_rate_cleaned.csv', 'date', ['UNRATE', 'unrate_yoy_change'])
sahm  = load_econ('../Data/Intermediate/SAHMCURRENT_cleaned.csv', 'observation_date', ['SAHMCURRENT'])

# Merge all economic indicators into one frame
econ = rec.merge(gdp, on='date', how='outer') \
         .merge(urate, on='date', how='outer') \
         .merge(sahm, on='date', how='outer') \
         .sort_values('date')

print(f"Economic data: {econ.shape[0]} months, {econ.shape[1]} columns")
print(f"Date range: {econ['date'].min().date()} → {econ['date'].max().date()}")
econ.head()
```

```python
# ── Build category monthly aggregates ─────────────────────────────────

cat_monthly = df.groupby(['date', 'category'])['search_interest'].mean().unstack('category')
cat_monthly.columns = [f'search_{c}' for c in cat_monthly.columns]

# Merge with economic data
merged = econ.merge(cat_monthly, left_on='date', right_index=True, how='inner')

print(f"Merged dataset: {merged.shape[0]} months × {merged.shape[1]} columns")
merged.head()
```

# 13a

```python
search_cols = [c for c in merged.columns if c.startswith('search_')]
econ_cols = ['USREC', 'gdp_yoy_pct_change', 'UNRATE', 'unrate_yoy_change', 'SAHMCURRENT']

corr = merged[search_cols + econ_cols].corr().loc[search_cols, econ_cols]
corr.index = [c.replace('search_', '').replace('_', ' ') for c in corr.index]

fig, ax = plt.subplots(figsize=(12, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r', center=0,
            linewidths=0.5, vmin=-0.7, vmax=0.7, ax=ax)
ax.set_title('Correlation: Category Search Interest vs Economic Indicators (Raw)')
plt.tight_layout()
plt.show()
```

# 13b

```python
def rolling_zscore(group, window=12):
    rolling_mean = group['search_interest'].rolling(window, min_periods=6).mean()
    rolling_std  = group['search_interest'].rolling(window, min_periods=6).std()
    group['z_score'] = (group['search_interest'] - rolling_mean) / rolling_std.clip(lower=0.1)
    return group

df_z = df.sort_values(['category', 'keyword', 'date']).copy()
df_z = df_z.groupby(['category', 'keyword'], group_keys=False).apply(rolling_zscore)
df_z = df_z.dropna(subset=['z_score'])

# Category-level detrended aggregates
cat_z_monthly = df_z.groupby(['date', 'category'])['z_score'].mean().unstack('category')
cat_z_monthly.columns = [f'zsearch_{c}' for c in cat_z_monthly.columns]

merged_z = econ.merge(cat_z_monthly, left_on='date', right_index=True, how='inner')

print(f"Detrended merged dataset: {merged_z.shape}")
```

# ── 13c. Correlation matrix — DETRENDED search vs economic indicators ───────


```python
zsearch_cols = [c for c in merged_z.columns if c.startswith('zsearch_')]

corr_z = merged_z[zsearch_cols + econ_cols].corr().loc[zsearch_cols, econ_cols]
corr_z.index = [c.replace('zsearch_', '').replace('_', ' ') for c in corr_z.index]

fig, axes = plt.subplots(1, 2, figsize=(22, 8))

# Raw
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r', center=0,
            linewidths=0.5, vmin=-0.7, vmax=0.7, ax=axes[0])
axes[0].set_title('Raw Search Interest vs Economic Indicators')

# Detrended
sns.heatmap(corr_z, annot=True, fmt='.2f', cmap='RdBu_r', center=0,
            linewidths=0.5, vmin=-0.7, vmax=0.7, ax=axes[1])
axes[1].set_title('Detrended (Z-Score) vs Economic Indicators')

plt.suptitle('Effect of Detrending on Correlations', fontsize=14)
plt.tight_layout()
plt.show()
```

# 13d

```python
# ── 13e. Lead/lag analysis — does search activity LEAD economic changes? ────
# This directly addresses your proposal: can search trends predict recessions?

lags = range(-6, 7)  # -6 = search leads by 6 months, +6 = search lags by 6

targets = ['UNRATE', 'gdp_yoy_pct_change', 'SAHMCURRENT']
lag_corrs = {}

for target in targets:
    results = {}
    for col in zsearch_cols:
        cat_name = col.replace('zsearch_', '').replace('_', ' ')
        rs = []
        for lag in lags:
            shifted = merged_z[col].shift(lag)
            r, _ = stats.pearsonr(shifted.dropna().values, 
                                   merged_z[target].loc[shifted.dropna().index].values)
            rs.append(r)
        results[cat_name] = rs
    lag_corrs[target] = pd.DataFrame(results, index=lags)

# Plot
fig, axes = plt.subplots(1, 3, figsize=(22, 7))
titles = {
    'UNRATE': 'Unemployment Rate',
    'gdp_yoy_pct_change': 'GDP YoY Growth',
    'SAHMCURRENT': 'Sahm Rule Indicator'
}

for i, target in enumerate(targets):
    lag_corrs[target].plot(ax=axes[i], linewidth=1.2, alpha=0.8)
    axes[i].axvline(0, color='black', linestyle='--', linewidth=0.8, alpha=0.5)
    axes[i].axhline(0, color='gray', linestyle='-', linewidth=0.5)
    axes[i].set_xlabel('Lag (months)\n← search LEADS economy | search LAGS economy →')
    axes[i].set_ylabel('Pearson r')
    axes[i].set_title(f'Search vs {titles[target]}')
    axes[i].legend(fontsize=7, loc='best')

plt.suptitle('Lead-Lag Correlations: Detrended Category Search vs Economic Indicators', fontsize=14)
plt.tight_layout()
plt.show()
```

# 13f

```python
rows = []
for target in targets:
    for cat in lag_corrs[target].columns:
        series = lag_corrs[target][cat]
        best_lag = series.abs().idxmax()
        best_r = series.loc[best_lag]
        rows.append({
            'category': cat,
            'indicator': titles[target],
            'optimal_lag_months': best_lag,
            'correlation_at_optimal': round(best_r, 3),
            'leads_or_lags': 'search LEADS' if best_lag < 0 else ('concurrent' if best_lag == 0 else 'search LAGS')
        })

lag_table = pd.DataFrame(rows)

# Show only cases where search leads the economy
leads = lag_table[lag_table['optimal_lag_months'] < 0].sort_values('correlation_at_optimal', key=abs, ascending=False)
print(f"Category-indicator pairs where search LEADS the economy ({len(leads)} of {len(lag_table)}):\n")
display(leads)
```

# 13g

```python
fig, axes = plt.subplots(3, 1, figsize=(18, 14), sharex=True)

rec_months = rec[rec['USREC'] == 1]['date']

# For each indicator, pick the category with strongest leading correlation
for i, target in enumerate(targets):
    lead_subset = leads[leads['indicator'] == titles[target]]
    if lead_subset.empty:
        continue
    best = lead_subset.iloc[0]
    cat_col = f"zsearch_{best['category'].replace(' ', '_')}"
    lag_n = int(best['optimal_lag_months'])
    
    ax1 = axes[i]
    ax2 = ax1.twinx()
    
    # Economic indicator
    ax1.plot(merged_z['date'], merged_z[target], color='black', linewidth=1.5, label=titles[target])
    ax1.set_ylabel(titles[target], color='black')
    
    # Search (shifted forward to show predictive alignment)
    ax2.plot(merged_z['date'], merged_z[cat_col].shift(-lag_n),
             color='darkorange', linewidth=1, alpha=0.8,
             label=f"{best['category']} (shifted {abs(lag_n)}mo forward)")
    ax2.set_ylabel('Search Z-Score (shifted)', color='darkorange')
    
    # Recession shading
    for d in rec_months:
        ax1.axvspan(d, d + pd.DateOffset(months=1), alpha=0.12, color='red')
    
    ax1.set_title(f"{best['category']} search leads {titles[target]} by {abs(lag_n)} months (r={best['correlation_at_optimal']:.2f})")
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left', fontsize=9)

axes[-1].set_xlabel('Date')
fig.suptitle('Best Leading Search Categories vs Economic Indicators\n(search shifted forward to show predictive alignment)', fontsize=14)
plt.tight_layout()
plt.show()
```
